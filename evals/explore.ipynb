# Exploring an eval run
Run one cell at a time: each cell's output appears below it, and every variable stays in memory for the next cell.

```python
from collections import Counter

from evals.harness import load_cases, run
from rag.pipeline import Index, embedder, load
```

```python
index = Index(load(), embedder())
cases = load_cases()
len(index.chunks), len(cases)
```

```python
results = await run(cases, index)          # retrieval only: no model call, free
```

```python
rows = [(case.category, result.id, result.retrieved) for case, result in zip(cases, results)]
rows
```

```python
found = Counter(category for category, _, retrieved in rows if retrieved)
total = Counter(category for category, _, retrieved in rows if retrieved is not None)
{category: f"{found[category]}/{total[category]}" for category in total}
```
